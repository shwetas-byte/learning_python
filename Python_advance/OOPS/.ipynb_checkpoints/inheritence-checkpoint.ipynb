## Single Inheritence 
### There is only one parent and one child class

```python
class parents:
    def st1(self):
        print("Parent")
class child(parents):
    def st2(self):
        print("Child")
```

```python
c=child()
c.st1()
c.st2()
```

## Multilevel Inheritence

```python
class grandfather:
    def g1(self):
        print("grandfather")
class father(grandfather):
    def f1(self):
        print("Father")
class child(father):
    def c1(self):
        print("child")
```

```python
c=child()
c.g1()
c.f1()
c.c1()
```

## Multiple Inheritence

```python
class tenth:
    def ten(self):
        print("tenth")
class twelve:
    def tw(self):
        print("twelve")
class grade(tenth,twelve):
    def gr(self):
        print("Graduate")
```

```python
g=grade()
g.ten()
g.tw()
g.gr()
```
