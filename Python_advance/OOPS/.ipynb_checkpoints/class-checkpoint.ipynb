```python
class cybrom:
    a=10
    b=20
```

```python
d=cybrom()
print(d.a + d.b)
```

```python
class cybrom():
    a=10    #classs variable
    b=20
    def myself(self):    # here self is object of class and self is used to detect the object of the class
        print(self.a + self.b)   # kyuki class ke variable hai function ke nhii
    def min(self,x,y):
        print(x-y)   # yha function ke variable hai inhe instance variable kehte haii
    def show():
        print("Welcome")
```

```python
d=cybrom()
d.myself()
d.min(3,4)
cybrom.show()   # class ke sath call kiya isliye self nhi likha to bhi chl gya
```
